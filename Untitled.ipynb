```python
import yfinance as yf

stocks = ["TCS.NS", "INFY.NS"]

for stock in stocks:
    data = yf.download(stock, period="5d")

    filename = stock.replace(".NS", "") + ".csv"

    data.to_csv(filename)

    print(f"Saved {filename}")
```

```python
url = "https://archives.nseindia.com/content/equities/EQUITY_L.csv"
df = pd.read_csv(url)
symbols = (df["SYMBOL"].head(100) + ".NS").tolist()
print(df[["SYMBOL", "NAME OF COMPANY"]].head(100))
```

```python
data = yf.download(
    tickers=symbols,
    period="1y",
    interval="1d",
    group_by='ticker',
    auto_adjust=True,
    threads=True
)
```

```python
filename = "stock_data_pipeline_v1.csv"
data.to_csv(filename)
```

```python
print(f"Stock data successfully saved as '{filename}'")
```
